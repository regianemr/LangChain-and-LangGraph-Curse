## Chat Básico

```python
client = OpenAI()
model = "gpt-3.5-turbo"

response = client.chat.completions.create(
  model=model,
  messages=[
    {
      "role": "user",
      "content": "Conte uma piada"
    }
  ]
)
```

```python
print(response)

print(type(response))
```

```python
message = (response.choices[0].message.content)
print(message)
```

## Roles

```python
response = client.chat.completions.create(
  model=model,
  message=[
    {"role": "system", "content": "You are a fictional investment assistant."},
    {"role": "user", "content": "What is the best low-risk investment you recommend for this year?"}
  ]
)
```

```python
message = (response.choices[0].message.content)
print(message)
```

```python
response = client.chat.completions.create(
  model=model,
  messages=[
    {"role": "assistant", "content": message},
    {"role": "user", "content": "What are the risks related to treasury bonds?"}
  ]
)
```

```python
print(response.choices[0].message.content)
```
