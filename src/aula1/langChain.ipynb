# Setpup

```python
from langchain_openai import OpenAI, ChatOpenAI
import os
import yaml
```

```python
with open('/home/regianemelo/Documentos/Curso LangChain/langchain/aula1/config.yaml', 'r') as config_file:
  config = yaml.safe_load(config_file)

os.environ['OPENAI_API_KEY'] = config['OPENAI_API_KEY']
```

# Completion

```python
openai = OpenAI(model_name='gpt-4o-mini')
frequency_penalty = 1
presence_penalty = 1
temperature = 1
max_tokens = 500
n = 1
seed = 123
response = openai.invoke(input='Quem foi Carl Sagan?', temperature=temperature, frequency_penalty=frequency_penalty, presence_penalty=presence_penalty, max_tokens=max_tokens, n=n, seed=seed)

print(response)
```

# Chat

```python
chat = ChatOpenAI(model_name='gpt-3.5-turbo')
messages = [
  {"role": "system", "content": "Você é um assistente que fornece informações sobre figuras histórias."},
  {"role": "user", "content": "Quem foi Carl Sagan?"}
]
response = chat.invoke(messages)
print(response)
```
